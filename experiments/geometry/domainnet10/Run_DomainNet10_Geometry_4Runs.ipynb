# DomainNet-10: four FedGAPS geometry runs

Dirichlet α = 0.1, 0.5 × seeds 42, 123. Each run exports initial raw, selected raw, and selected smoothed prototypes. Extract the whole package first.

```python
from pathlib import Path
import subprocess, sys, os
PROJECT = Path.cwd().resolve()
required = ["run_geometry.py", "icde/__init__.py", "icde/config.py", "reference/run_digits5.py"]
missing = [name for name in required if not (PROJECT / name).is_file()]
assert not missing, f"Extract the ENTIRE ZIP and open its notebook. Project: {PROJECT}; missing: {missing}"
ENV = dict(os.environ)
ENV["PYTHONPATH"] = str(PROJECT) + os.pathsep + ENV.get("PYTHONPATH", "")
subprocess.run([sys.executable, "-c", "from icde.config import Config; import icde.config; print('Config loaded:', icde.config.__file__)"], cwd=PROJECT, env=ENV, check=True)
candidates = [PROJECT / "data", PROJECT.parent / "data"]
DATA_ROOT = next((p for p in candidates if (p / "domainnet10").is_dir()), PROJECT / "data")
# You can set DATA_ROOT explicitly to the existing directory containing domainnet10.
OUTPUT = PROJECT / "outputs" / "geometry_4runs"
RESUME = True
WORKERS = 4
BATCH_SIZE = 32
AMP = True
print("Data root:", DATA_ROOT)
print("Output:", OUTPUT)
```

```python
args = [sys.executable, str(PROJECT / "run_geometry.py"), "--data-root", str(DATA_ROOT), "--output", str(OUTPUT), "--device", "cuda", "--workers", str(WORKERS), "--batch-size", str(BATCH_SIZE)]
if AMP: args.append("--amp")
if RESUME: args.append("--resume")
subprocess.run(args, cwd=PROJECT, env=ENV, check=True)
```

```python
files = [OUTPUT / "geometry_per_run.csv", OUTPUT / "geometry_table.tex"]
if all(p.is_file() for p in files):
    for p in files:
        print(p.read_text())
else:
    print("Results are not ready. All four runs and the summary must finish successfully first. Check the run cell above.")
```
