# Main benchmark: baselines and ablations
Run this notebook from the fully extracted repository. Planning does not start training. Sensitivity notebooks are under `experiments/sensitivity/`.

```python
from pathlib import Path
import os, sys, subprocess
PROJECT = Path.cwd().resolve()
assert (PROJECT / "icde/config.py").is_file(), "Open this notebook from the repository root."
DATASET = "digits5"  # digits5, office10, domainnet10
DATA_ROOT = PROJECT / "data"  # Parent of the dataset folders
OUTPUT = PROJECT / "outputs" / f"{DATASET}_benchmark"
CONFIG = PROJECT / "configs" / f"{DATASET}.json"
DEVICE = "cuda"
ENV = dict(os.environ)
ENV["PYTHONPATH"] = str(PROJECT) + os.pathsep + ENV.get("PYTHONPATH", "")
def run(*args):
    subprocess.run([sys.executable, *map(str,args)], cwd=PROJECT, env=ENV, check=True)
```

```python
import torch
print("PyTorch:", torch.__version__, "CUDA:", torch.cuda.is_available())
if DEVICE == "cuda":
    assert torch.cuda.is_available(), "Choose DEVICE=cpu or use a CUDA-enabled environment."
    print(torch.cuda.get_device_name(0))
```

```python
args = ["-m", "icde", "sweep", "--config", CONFIG, "--ablations",
        "--data-root", DATA_ROOT, "--output-root", OUTPUT, "--device", DEVICE]
run(*args)
```

## Run training
The supplied configuration plans 90 runs: 15 methods × two concentrations × three seeds. Change the configuration before planning if a different seed set is needed. Keep the same configuration and source when resuming.

```python
assert (DATA_ROOT / DATASET).is_dir(), f"Missing dataset: {DATA_ROOT / DATASET}"
run(*args, "--execute", "--resume")
```
